**Blok tabel sumber 7** · diekstrak persis dari tabel Word Master R53; bab 1.

```python
S = 100
bandwidth = [10, 20, 50, 100]

hasil = []

for B in bandwidth:
    T = S / B
    hasil.append(T)

for B, T in zip(bandwidth, hasil):
    print(f"Bandwidth = {B:3} MB/s, "
          f"Waktu = {T:5.2f} s")
```

**Blok tabel sumber 8** · diekstrak persis dari tabel Word Master R53; bab 1.

```python
import matplotlib.pyplot as plt

plt.plot(bandwidth, hasil, marker="o")

plt.xlabel("Bandwidth (MB/s)")
plt.ylabel("Waktu Transfer (s)")
plt.title("Hubungan Bandwidth dan Waktu Transfer")
plt.grid(True)

plt.show()
```
