**Blok sumber 3** · gaya Word `Code` · paragraf sekitar 451.

```python
import math

x_exact = math.sqrt(2)
x_approx = 1.4142

absolute_error = abs(x_exact - x_approx)
relative_error = absolute_error / abs(x_exact)
relative_error_percent = relative_error * 100

print("Galat absolut :", absolute_error)
print("Galat relatif :", relative_error)
print("Galat relatif (%) :", relative_error_percent)
```

**Blok sumber 8** · gaya Word `Code` · paragraf sekitar 541.

```python
import math

x = 0.5
exact = math.exp(x)

for n in range(6):
    approximation = sum(
        x**k / math.factorial(k)
        for k in range(n + 1)
    )

    error = abs(exact - approximation)

    print(n, approximation, error)
```

**Blok sumber 10** · gaya Word `Code` · paragraf sekitar 612.

```python
def f(x):
    return 1 / (1 - x)

x = 0.999
dx = 1e-6

y1 = f(x)
y2 = f(x + dx)

relative_input_change = abs(dx / x)
relative_output_change = abs((y2 - y1) / y1)

print("f(x) =", y1)
print("f(x + dx) =", y2)
print("Perubahan relatif input =", relative_input_change)
print("Perubahan relatif output =", relative_output_change)
```
