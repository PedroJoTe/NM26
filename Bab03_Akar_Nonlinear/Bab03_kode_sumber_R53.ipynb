**Blok sumber 2** · gaya Word `CodeBlock` · paragraf sekitar 960.

````python
def bisection(f, a, b, eps_x=1e-6, eps_f=1e-6, n_max=100):
    fa = f(a)
    fb = f(b)

    if fa == 0:
        return a, [], "converged"

    if fb == 0:
        return b, [], "converged"

    if fa * fb > 0:
        raise ValueError("Interval tidak memenuhi syarat bracketing.")

    history = []
    previous = None

    for n in range(n_max):
        x = (a + b) / 2
        fx = f(x)
        residual = abs(fx)

        if previous is None:
            delta = None
        else:
            delta = abs(x - previous)

        history.append({
            "n": n,
            "x_n": x,
            "delta": delta,
            "residual": residual
        })

        if fx == 0:
            return x, history, "converged"

        if delta is not None:
            if delta <= eps_x and residual <= eps_f:
                return x, history, "converged"

        if fa * fx < 0:
            b = x
            fb = fx
        else:
            a = x
            fa = fx

        previous = x

    return x, history, "maximum iteration reached"
````

````python
def regula_falsi(f, a, b, eps_x=1e-6, eps_f=1e-6, n_max=100):
    fa = f(a)
    fb = f(b)

    if fa == 0:
        return a, [], "converged"

    if fb == 0:
        return b, [], "converged"

    if fa * fb > 0:
        raise ValueError("Interval tidak memenuhi syarat bracketing.")

    history = []
    previous = None

    for n in range(n_max):
        denominator = fb - fa

        if abs(denominator) < 1e-14:
            raise ZeroDivisionError(
                "Denominator Regula Falsi terlalu dekat dengan nol."
            )

        x = (a * fb - b * fa) / denominator
        fx = f(x)
        residual = abs(fx)

        if previous is None:
            delta = None
        else:
            delta = abs(x - previous)

        history.append({
            "n": n,
            "x_n": x,
            "delta": delta,
            "residual": residual
        })

        if fx == 0:
            return x, history, "converged"

        if delta is not None:
            if delta <= eps_x and residual <= eps_f:
                return x, history, "converged"

        if fa * fx < 0:
            b = x
            fb = fx
        else:
            a = x
            fa = fx

        previous = x

    return x, history, "maximum iteration reached"
````

**Blok sumber 11** · gaya Word `CodeBlock` · paragraf sekitar 1328.

````python
def newton_raphson(f, df, x0, eps_x=1e-6, eps_f=1e-6, n_max=100):
    history = []
    x = x0

    if f(x) == 0:
        return x, history, "converged"

    for n in range(1, n_max + 1):
        fx = f(x)
        dfx = df(x)

        if abs(dfx) < 1e-14:
            raise ZeroDivisionError(
                "Turunan terlalu dekat dengan nol."
            )

        x_new = x - fx / dfx

        delta = abs(x_new - x)
        residual = abs(f(x_new))

        history.append({
            "n": n,
            "x_n": x_new,
            "delta": delta,
            "residual": residual
        })

        if residual == 0:
            return x_new, history, "converged"

        if delta <= eps_x and residual <= eps_f:
            return x_new, history, "converged"

        x = x_new

    return x, history, "maximum iteration reached"
````

**Blok sumber 16** · gaya Word `CodeBlock` · paragraf sekitar 1499.

````python
def secant(f, x0, x1, eps_x=1e-6, eps_f=1e-6, n_max=100):
    history = []

    if f(x0) == 0:
        return x0, history, "converged"

    if f(x1) == 0:
        return x1, history, "converged"

    for n in range(2, n_max + 2):
        f0 = f(x0)
        f1 = f(x1)

        denominator = f1 - f0

        if abs(denominator) < 1e-14:
            raise ZeroDivisionError(
                "Denominator Secant terlalu dekat dengan nol."
            )

        x2 = x1 - f1 * (x1 - x0) / denominator

        delta = abs(x2 - x1)
        residual = abs(f(x2))

        history.append({
            "n": n,
            "x_n": x2,
            "delta": delta,
            "residual": residual
        })

        if residual == 0:
            return x2, history, "converged"

        if delta <= eps_x and residual <= eps_f:
            return x2, history, "converged"

        x0, x1 = x1, x2

    return x1, history, "maximum iteration reached"
````

````python
def server_response(lambda_rate):
    return (
        1 / (100 - lambda_rate)
        + 1e-6 * lambda_rate**2
    )

def server_equation(lambda_rate):
    return server_response(lambda_rate) - 0.05
````
