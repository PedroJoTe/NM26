```python
def linear_interpolation(x0, y0, x1, y1, x):
    if x0 == x1:
        raise ValueError("x0 dan x1 harus berbeda.")

    y = y0 + (y1 - y0) / (x1 - x0) * (x - x0)
    return y


x0, y0 = 2, 4
x1, y1 = 5, 10
x = 3.5

hasil = linear_interpolation(x0, y0, x1, y1, x)

print(f"Hasil interpolasi = {hasil:.2f}")
```

```python
import numpy as np

def lagrange_interpolation(x_data, y_data, x):
    x_data = np.asarray(x_data, dtype=float)
    y_data = np.asarray(y_data, dtype=float)

    if len(x_data) != len(y_data):
        raise ValueError("Jumlah x dan y harus sama.")

    if len(np.unique(x_data)) != len(x_data):
        raise ValueError("Node x tidak boleh duplikat.")

    p = 0.0
    n = len(x_data)

    for i in range(n):
        L = 1.0

        for j in range(n):
            if i != j:
                L *= (x - x_data[j]) / (x_data[i] - x_data[j])

        p += y_data[i] * L

    return p


x_data = [1, 2, 4]
y_data = [1, 4, 16]

hasil = lagrange_interpolation(x_data, y_data, 3)

print(f"P(3) = {hasil:.2f}")
```

```python
import numpy as np

def newton_divided_difference(x_data, y_data, x):
    x_data = np.asarray(x_data, dtype=float)
    y_data = np.asarray(y_data, dtype=float)

    if len(x_data) != len(y_data):
        raise ValueError("Jumlah x dan y harus sama.")

    if len(np.unique(x_data)) != len(x_data):
        raise ValueError("Node x tidak boleh duplikat.")

    n = len(x_data)
    table = np.zeros((n, n))
    table[:, 0] = y_data

    for j in range(1, n):
        for i in range(n - j):
            table[i, j] = (
                table[i + 1, j - 1] - table[i, j - 1]
            ) / (x_data[i + j] - x_data[i])

    p = table[0, 0]
    product = 1.0

    for j in range(1, n):
        product *= x - x_data[j - 1]
        p += table[0, j] * product

    return p, table


x_data = [1, 2, 4]
y_data = [1, 4, 16]

hasil, table = newton_divided_difference(
    x_data, y_data, 3
)

print(f"P(3) = {hasil:.2f}")
print(table)
```

```python
import numpy as np

def natural_cubic_spline(x_data, y_data, x_target):
    x_data = np.asarray(x_data, dtype=float)
    y_data = np.asarray(y_data, dtype=float)

    n = len(x_data)

    if n != len(y_data):
        raise ValueError("Jumlah x dan y harus sama.")

    if n < 3:
        raise ValueError("Minimal diperlukan tiga titik.")

    if np.any(np.diff(x_data) <= 0):
        raise ValueError("x_data harus meningkat dan unik.")

    h = np.diff(x_data)

    # Sistem untuk M₁, ..., M_(n-1)
    A = np.zeros((n - 2, n - 2))
    b = np.zeros(n - 2)

    for i in range(1, n - 1):
        row = i - 1

        if row > 0:
            A[row, row - 1] = h[i - 1]

        A[row, row] = 2 * (h[i - 1] + h[i])

        if row < n - 3:
            A[row, row + 1] = h[i]

        b[row] = 6 * (
            (y_data[i + 1] - y_data[i]) / h[i]
            - (y_data[i] - y_data[i - 1]) / h[i - 1]
        )

    M = np.zeros(n)

    # np.linalg.solve menyelesaikan SPL yang dibentuk
    # dari kondisi natural cubic spline.
    M[1:n - 1] = np.linalg.solve(A, b)

    i = np.searchsorted(x_data, x_target) - 1
    i = np.clip(i, 0, n - 2)

    hi = h[i]
    dx = x_target - x_data[i]

    a = y_data[i]
    b_coef = (
        (y_data[i + 1] - y_data[i]) / hi
        - hi * (2 * M[i] + M[i + 1]) / 6
    )
    c = M[i] / 2
    d = (M[i + 1] - M[i]) / (6 * hi)

    return a + b_coef * dx + c * dx**2 + d * dx**3


x_data = [0, 1, 2, 3]
y_data = [1, 2, 1, 2]

hasil = natural_cubic_spline(x_data, y_data, 1.5)

print(f"S(1.5) = {hasil:.2f}")
```

```python
import numpy as np

def linear_regression(x_data, y_data):
    x = np.asarray(x_data, dtype=float)
    y = np.asarray(y_data, dtype=float)

    n = len(x)

    A = np.array([
        [n, np.sum(x)],
        [np.sum(x), np.sum(x**2)]
    ])

    b = np.array([
        np.sum(y),
        np.sum(x * y)
    ])

    a, slope = np.linalg.solve(A, b)

    y_pred = a + slope * x
    residual = y - y_pred

    sse = np.sum(residual**2)
    rmse = np.sqrt(np.mean(residual**2))

    return a, slope, y_pred, residual, sse, rmse


x = [1, 2, 3, 4]
y = [2.2, 2.9, 3.8, 5.1]

a, b, y_pred, residual, sse, rmse = \
    linear_regression(x, y)

print(f"a = {a:.2f}")
print(f"b = {b:.2f}")
print(f"SSE = {sse:.3f}")
print(f"RMSE = {rmse:.4f}")
```

```python
import numpy as np

x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([2.2, 2.9, 3.8, 5.1], dtype=float)

# Matriks desain A = [1, x, x^2]
A = np.column_stack((
    np.ones_like(x),
    x,
    x**2
))

# Persamaan normal:
# A^T A beta = A^T y
ATA = A.T @ A
ATy = A.T @ y

beta = np.linalg.solve(ATA, ATy)

a0, a1, a2 = beta

y_pred = A @ beta
residual = y - y_pred

sse = np.sum(residual**2)
rmse = np.sqrt(np.mean(residual**2))

print(f"a0 = {a0:.2f}")
print(f"a1 = {a1:.2f}")
print(f"a2 = {a2:.2f}")
print(f"SSE = {sse:.3f}")
print(f"RMSE = {rmse:.4f}")
```
