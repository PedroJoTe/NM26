**Blok sumber 4** · gaya Word `Code` · paragraf sekitar 3323.

````python
def forward_difference(f, x, h):
    return (f(x + h) - f(x)) / h


def backward_difference(f, x, h):
    return (f(x) - f(x - h)) / h


def central_difference(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)
````

````python
import numpy as np
import matplotlib.pyplot as plt

x = 1.0
h_values = 10.0 ** (-np.arange(1, 7))

exact = np.cos(x)
errors = []

for h in h_values:
    numerical = central_difference(np.sin, x, h)
    error = abs(exact - numerical)
    errors.append(error)

plt.loglog(h_values, errors, marker="o")
plt.xlabel("h")
plt.ylabel("Galat absolut")
plt.title("Galat Beda Tengah terhadap Ukuran Langkah")
plt.grid(True, which="both")
plt.show()
````

**Blok sumber 10** · gaya Word `Code` · paragraf sekitar 3429.

````python
def trapezoidal_rule(f, a, b, n):
    if not isinstance(n, int) or isinstance(n, bool) or n <= 0:
        raise ValueError("n harus berupa bilangan bulat positif")

    h = (b - a) / n
    total = f(a) + f(b)

    for i in range(1, n):
        x = a + i * h
        total += 2 * f(x)

    return (h / 2) * total
````

**Blok sumber 11** · gaya Word `Code` · paragraf sekitar 3431.

````python
def simpson_one_third(f, a, b, n):
    if not isinstance(n, int) or isinstance(n, bool) or n <= 0:
        raise ValueError("n harus berupa bilangan bulat positif")
    if n % 2 != 0:
        raise ValueError("n harus genap untuk Simpson 1/3")

    h = (b - a) / n
    total = f(a) + f(b)

    for i in range(1, n):
        x = a + i * h

        if i % 2 == 0:
            total += 2 * f(x)
        else:
            total += 4 * f(x)

    return (h / 3) * total
````

**Blok sumber 16** · gaya Word `Code` · paragraf sekitar 3500.

````python
def trapezoidal_data(y, h):
    if len(y) < 2:
        raise ValueError("Data harus memuat minimal dua titik")
    if h <= 0:
        raise ValueError("h harus positif")

    total = y[0] + y[-1]

    for i in range(1, len(y) - 1):
        total += 2 * y[i]

    return (h / 2) * total


def simpson_data(y, h):
    if len(y) < 3:
        raise ValueError("Simpson 1/3 membutuhkan minimal tiga titik data")
    if h <= 0:
        raise ValueError("h harus positif")

    n = len(y) - 1
    if n % 2 != 0:
        raise ValueError(
            "Simpson 1/3 membutuhkan jumlah subinterval genap."
        )

    total = y[0] + y[-1]

    for i in range(1, n):
        if i % 2 == 0:
            total += 2 * y[i]
        else:
            total += 4 * y[i]

    return (h / 3) * total
````
