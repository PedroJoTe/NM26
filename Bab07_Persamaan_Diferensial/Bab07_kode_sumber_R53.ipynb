**Blok sumber 2** · gaya Word `Normal` · paragraf sekitar 3724.

````python
def euler_ode(f, t0, y0, tf, h):
    if h <= 0 or tf < t0:
        raise ValueError("h harus positif dan tf tidak boleh lebih kecil dari t0")

    t = [float(t0)]
    y = [float(y0)]

    time_tol = 1e-12 * max(1.0, abs(tf))
    while t[-1] < tf - time_tol:
        tn = t[-1]
        yn = y[-1]
        remaining = tf - tn
        h_step = min(h, remaining)

        ynext = yn + h_step * f(tn, yn)
        tnext = tf if remaining <= h + time_tol else tn + h_step

        t.append(tnext)
        y.append(ynext)

    return t, y
````

**Blok sumber 9** · gaya Word `Normal` · paragraf sekitar 3946.

````python
def rk4_ode(f, t0, y0, tf, h):
    if h <= 0 or tf < t0:
        raise ValueError("h harus positif dan tf tidak boleh lebih kecil dari t0")

    t = [float(t0)]
    y = [float(y0)]

    time_tol = 1e-12 * max(1.0, abs(tf))
    while t[-1] < tf - time_tol:
        tn = t[-1]
        yn = y[-1]
        remaining = tf - tn
        h_step = min(h, remaining)

        k1 = f(tn, yn)
        k2 = f(tn + h_step / 2, yn + h_step * k1 / 2)
        k3 = f(tn + h_step / 2, yn + h_step * k2 / 2)
        k4 = f(tn + h_step, yn + h_step * k3)

        ynext = yn + h_step * (k1 + 2*k2 + 2*k3 + k4) / 6
        tnext = tf if remaining <= h + time_tol else tn + h_step

        t.append(tnext)
        y.append(ynext)

    return t, y
````
