**Blok sumber 2** · gaya Word `Code` · paragraf sekitar 4468.

````python
import math

def golden_section(f, a, b, tol, max_iter):
    r = (math.sqrt(5) - 1) / 2

    x1 = b - r * (b - a)
    x2 = a + r * (b - a)

    f1 = f(x1)
    f2 = f(x2)

    history = []
    converged = False

    for k in range(1, max_iter + 1):

        width = abs(b - a)

        history.append({
            "iteration": k,
            "a": a,
            "b": b,
            "x1": x1,
            "x2": x2,
            "f_x1": f1,
            "f_x2": f2,
            "width": width
        })

        if width <= tol:
            converged = True
            break

        if f1 > f2:
            a = x1
            x1 = x2
            f1 = f2

            x2 = a + r * (b - a)
            f2 = f(x2)

        else:
            b = x2
            x2 = x1
            f2 = f1

            x1 = b - r * (b - a)
            f1 = f(x1)

    x_opt = (a + b) / 2

    return {
        "x_opt": x_opt,
        "f_opt": f(x_opt),
        "iterations": len(history),
        "history": history,
        "converged": converged
    }
````

**Blok sumber 7** · gaya Word `Code` · paragraf sekitar 4627.

````python
def newton_optimize(f, df, d2f, x0, tol, max_iter):
    x = float(x0)

    history = []
    converged = False

    for k in range(1, max_iter + 1):

        d1 = df(x)
        d2 = d2f(x)

        if abs(d2) < 1e-14:
            break

        x_new = x - d1 / d2

        f_new = f(x_new)

        dx = abs(x_new - x)
        df_value = abs(f_new - f(x))

        history.append({
            "iteration": k,
            "x_old": x,
            "x_new": x_new,
            "f_new": f_new,
            "delta_x": dx,
            "delta_f": df_value,
            "df": d1,
            "d2f": d2
        })

        x = x_new

        if dx <= tol:
            converged = True
            break

    return {
        "x_opt": x,
        "f_opt": f(x),
        "iterations": len(history),
        "history": history,
        "converged": converged
    }
````

ALGORITMA GradientDescent

Input:
    fungsi J(x)
    gradient grad_J(x)
    initial point x0
    learning rate alpha
    toleransi epsilon_g
    maksimum iterasi n_max

    hitung:
        norm_g = ||g||

        berhenti

**Blok sumber 14** · gaya Word `Code` · paragraf sekitar 4814.

````python
import numpy as np

def gradient_descent(f, grad_f, x0, alpha, tol, max_iter):
    x = np.asarray(x0, dtype=float)

    history = []
    converged = False

    for k in range(1, max_iter + 1):

        g = np.asarray(grad_f(x), dtype=float)
        grad_norm = np.linalg.norm(g)

        history.append({
            "iteration": k,
            "x": x.copy(),
            "f": f(x),
            "gradient_norm": grad_norm
        })

        if grad_norm <= tol:
            converged = True
            break

        x = x - alpha * g

    return {
        "x_opt": x,
        "f_opt": f(x),
        "iterations": len(history),
        "history": history,
        "converged": converged}
````
